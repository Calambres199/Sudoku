# Ejemplos de uso del paquete `sudoku_pkg`

Esta libreta muestra un ejemplo práctico del uso del código, el cual resuelve Sudokus y determina si la solución es única o no.

El paquete implementa un resolvedor de Sudoku 9×9 utilizando el algoritmo de **backtracking** (búsqueda con retroceso).

## 1. Instalación del paquete

Primero, clonamos el repositorio e instalamos el paquete.

```python
!git clone https://github.com/Calambres199/Sudoku
%cd Sudoku
!pip install .
```

## 2. Importar las funciones necesarias

```python
from sudoku_pkg.sudoku import (
    es_valido,
    solve_sudoku,
    there_is_another_solution,
    print_grid
)
```

## 3. Definir un Sudoku de ejemplo

Los ceros representan las celdas vacías.

```python
sudoku = [
    [5,3,0,0,7,0,0,0,0],
    [6,0,0,1,9,5,0,0,0],
    [0,9,8,0,0,0,0,6,0],
    [8,0,0,0,6,0,0,0,3],
    [4,0,0,8,0,3,0,0,1],
    [7,0,0,0,2,0,0,0,6],
    [0,6,0,0,0,0,2,8,0],
    [0,0,0,4,1,9,0,0,5],
    [0,0,0,0,8,0,0,7,9]
]
```

## 4. Guardar una copia del original

Esto es necesario para verificar si la solución es única más adelante.

```python
original = [fila[:] for fila in sudoku]
```

## 5. Verificar si el Sudoku es válido

La función `es_valido` revisa que no haya repeticiones en filas, columnas o subcuadrículas.

```python
es_valido(sudoku)
```

## 6. Resolver el Sudoku

La función `solve_sudoku` aplica backtracking para encontrar la solución.

```python
solve_sudoku(sudoku)
```

## 7. Mostrar la solución

La función `print_grid` imprime el tablero de forma legible.

```python
print_grid(sudoku)
```

## 8. Verificar si la solución es única

La función `there_is_another_solution` determina si el Sudoku tiene más de una solución.

```python
there_is_another_solution(original, sudoku)
```

## 9. Probar con un Sudoku inválido (repetición en fila)

```python
sudoku_invalido_fila = [
    [5,5,0,0,7,0,0,0,0],
    [6,0,0,1,9,5,0,0,0],
    [0,9,8,0,0,0,0,6,0],
    [8,0,0,0,6,0,0,0,3],
    [4,0,0,8,0,3,0,0,1],
    [7,0,0,0,2,0,0,0,6],
    [0,6,0,0,0,0,2,8,0],
    [0,0,0,4,1,9,0,0,5],
    [0,0,0,0,8,0,0,7,9]
]
es_valido(sudoku_invalido_fila)
```

## 10. Probar con un Sudoku inválido (repetición en columna)

```python
sudoku_invalido_columna = [
    [5,3,0,0,7,0,0,0,0],
    [5,0,0,1,9,5,0,0,0],
    [0,9,8,0,0,0,0,6,0],
    [8,0,0,0,6,0,0,0,3],
    [4,0,0,8,0,3,0,0,1],
    [7,0,0,0,2,0,0,0,6],
    [0,6,0,0,0,0,2,8,0],
    [0,0,0,4,1,9,0,0,5],
    [0,0,0,0,8,0,0,7,9]
]
es_valido(sudoku_invalido_columna)
```

## 11. Probar con un Sudoku inválido (repetición en subcuadrícula)

```python
sudoku_invalido_caja = [
    [5,3,0,0,7,0,0,0,0],
    [6,5,0,1,9,5,0,0,0],
    [0,9,8,0,0,0,0,6,0],
    [8,0,0,0,6,0,0,0,3],
    [4,0,0,8,0,3,0,0,1],
    [7,0,0,0,2,0,0,0,6],
    [0,6,0,0,0,0,2,8,0],
    [0,0,0,4,1,9,0,0,5],
    [0,0,0,0,8,0,0,7,9]
]
es_valido(sudoku_invalido_caja)
```
